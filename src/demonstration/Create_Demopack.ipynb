# Create a demopack

This notebook copies a demonstration run into a portable demopack. Review the source and target first: exporting can replace an existing demopack.

```python
import sys
sys.path.append('..')

from pathlib import Path
import shutil

from exp_run_config import Config

Config.PROJECTNAME = 'BerryPicker'
```

## Choose the source and target

Set the demonstration run to package and the name of the demopack to create.

```python
demo_experiment = 'demonstration'
demo_run = 'random-both-cameras-video'
demopack_name = 'random-both-cameras-video'

source_path = Path(Config()['experiment_data'], demo_experiment, demo_run).expanduser()
target_path = Path(Config()['demopacks_path'], demopack_name).expanduser()
```

## Preview the export

This cell is read-only. Verify the source demonstrations and target location before enabling export.

```python
source_demos = sorted(path.name for path in source_path.iterdir() if path.is_dir()) if source_path.exists() else []
print(f'Source: {source_path}')
print(f'Source exists: {source_path.exists()}')
print(f'Source demonstrations ({len(source_demos)}): {source_demos}')
print(f'Target: {target_path}')
print(f'Target exists: {target_path.exists()}')
```

## Export deliberately

Set `EXPORT_DEMOPACK` to `True` to copy the source. If the target already exists, also set `OVERWRITE_EXISTING` to `True` to replace it.

```python
EXPORT_DEMOPACK = False
OVERWRITE_EXISTING = False
```

```python
assert EXPORT_DEMOPACK, 'Set EXPORT_DEMOPACK = True after reviewing the preview above.'
assert source_path.exists(), f'Source does not exist: {source_path}'
if target_path.exists():
    assert OVERWRITE_EXISTING, 'Set OVERWRITE_EXISTING = True to replace the existing target.'
    shutil.rmtree(target_path)

target_path.parent.mkdir(parents=True, exist_ok=True)
shutil.copytree(source_path, target_path)
print(f'Created demopack: {target_path}')
```

## Verify the exported pack

Run this after export to confirm that the expected demonstration directories were copied.

```python
pack_demos = sorted(path.name for path in target_path.iterdir() if path.is_dir())
print(f'Demopack: {target_path}')
print(f'Packaged demonstrations ({len(pack_demos)}): {pack_demos}')
```
